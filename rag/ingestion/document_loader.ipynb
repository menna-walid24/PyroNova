# Document loader
Step 1 rebuilds `knowledge_base/experiments/` from the **real NASA PSI-159 CFI-G data** (`Data/raw/psi159_experiments.json`). Step 2 loads the Markdown documents.

```python
from pathlib import Path
import sys

# Works on every machine: find the project root (the folder that contains rag/pipeline.py)
PROJECT_ROOT = next(
    p for p in [Path.cwd().resolve(), *Path.cwd().resolve().parents]
    if (p / "rag" / "pipeline.py").exists()
)
RAG_DIR = PROJECT_ROOT / "rag"

INGESTION_DIR = RAG_DIR / "ingestion"
if str(INGESTION_DIR) not in sys.path:
    sys.path.insert(0, str(INGESTION_DIR))

import importlib
import json_to_documents
importlib.reload(json_to_documents)

created = json_to_documents.main()
assert created == 179, f"expected 179 real tests, got {created}"
```

```python
KNOWLEDGE_BASE = PROJECT_ROOT / "knowledge_base" / "experiments"


def load_documents():
    documents = []
    for file_path in sorted(KNOWLEDGE_BASE.glob("*.md")):
        with open(file_path, "r", encoding="utf-8") as f:
            content = f.read()
        documents.append({"text": content, "source": str(file_path)})
    return documents


documents = load_documents()
print(f"Loaded {len(documents)} documents")

old = [d for d in documents if "EXP-" in Path(d["source"]).name]
assert not old, "old synthetic EXP documents are still in the knowledge base"

if documents:
    print("\nFirst document:")
    print(documents[0]["text"][:1000])
```
